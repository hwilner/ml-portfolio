Q1.b Compute the average from the sum of the derivatives of the error of all points.

````
def error (a,b,X,y):
  return .5 * (y - (a * X + b)) ** 2

# calculatin 'a' derivative error in dicrit point 
def a_error (a,b,X,y):
  return (y-(a*X+b))*(-X)

# calculatin 'b' derivative error in dicrit point             
def b_error (a,b,X,y):
  return (y-a*X-b)*(-1)

#updating a and b by the derivative mean
def updaitig_weigth (a,b,x_nor,y_nor,learning_rate):
  new_a = a - learning_rate * mean([a_error(a,b,x,y) for x,y in zip(x_nor,y_nor) ])
  new_b = b - learning_rate * mean([b_error(a,b,x,y) for x,y in zip(x_nor,y_nor) ])
  cor_error = mean([error(a,b,x,y) for x,y in zip(x_nor,y_nor) ])
  return float(new_a),float(new_b),float(cor_error)


# function for averaging list
def mean(a):
  return sum(a)/float(len(a))

## trying to understand error in the while loop
d,f,g=updaitig_weigth (a,b,x_nor,y_nor,0.002) 
print(d,f,g)
print(type(d),type(f),type(g))
````

````

def luhn_is_valid(card_number: str) -> bool:
    """Validate a card number with the Luhn (mod-10) checksum.

    Starting from the rightmost digit, every second digit is doubled and any
    result above nine has nine subtracted. The final sum of digits must be
    divisible by ten.

    Args:
        card_number: Card number as a string. Spaces and dashes are ignored.

    Returns:
        bool: ``True`` if the checksum passes, ``False`` otherwise.

    Raises:
        ValueError: If the input is empty or contains characters other than
            digits, spaces and dashes.

    Example:
        >>> luhn_is_valid("4799 2739 8713 6272")
        True
        >>> luhn_is_valid("1234")
        False
    """
    digits = [int(ch) for ch in card_number.replace(" ", "").replace("-", "")]
    if not digits:
        raise ValueError("card_number must not be empty")

    checksum = 0
    for position, digit in enumerate(reversed(digits)):
        if position % 2 == 1:
            digit *= 2
            if digit > 9:          # doubled 7-9 wrap around by subtracting 9
                digit -= 9
        checksum += digit
    return checksum % 10 == 0


# The original cell called input(), which blocks during non-interactive
# execution (nbconvert, CI, a fresh clone). Testing known-good numbers keeps
# the exercise reproducible.
for number, expected in [("4799 2739 8713 6272", True), ("4539 1488 0343 6467", True),
                         ("1234 5678 9012 3456", False), ("7992 7391 4430", True)]:
    result = luhn_is_valid(number)
    flag = "OK" if result == expected else "MISMATCH"
    print(f"{number:22} -> {result}  (expected {expected})  {flag}")
````
